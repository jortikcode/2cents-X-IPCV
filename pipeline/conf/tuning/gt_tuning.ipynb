# Tuning Interattivo Ground Truth (gt.csv)

Questo notebook interattivo consente di ispezionare visivamente e correggere con precisione la posizione dei centri $(cx, cy)$ e i raggi $r$ di ogni moneta annotata nel dataset target (`pipeline/conf/gt.csv`).

### Caratteristiche principali:
- **Navigazione Rapida**: Menu a tendina per selezionare ogni immagine del dataset (con pulsanti `◀ Prec` e `Succ ▶`).
- **Selezione Moneta**: Nelle immagini con più monete, un selettore consente di scegliere quale specifica moneta modificare.
- **Doppia Vista Sincronizzata**:
  1. *Panoramica completa*: Mostra l'intera immagine con tutte le monete (quella attiva in verde brillante con mirino rosso; le altre in ciano tratteggiato).
  2. *Zoom Pixel-Perfect*: Crop ad alta risoluzione centrato sulla moneta selezionata con cerchi guida di precisione ($r \pm 2\text{ px}$) per allineare perfettamente il contorno al bordo fisico della moneta.
- **Controlli Fine-Tuning**: Slider continui e pulsanti rapidi step-by-step ($\pm 1\text{ px}$ e $\pm 5\text{ px}$) per $cx$, $cy$ e raggio $r$.
- **Salvataggio Istantaneo**: Il pulsante **Salva Modifiche** aggiorna immediatamente la riga corrispondente nel file `pipeline/conf/gt.csv`.

```python
import sys
from pathlib import Path
from IPython.display import display

# Risoluzione dinamica della root del progetto (relativa)
current_dir = Path.cwd().resolve()
project_root = current_dir
for parent in [current_dir] + list(current_dir.parents):
    if (parent / 'coin_dataset' / 'target_set').exists():
        project_root = parent
        break

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from pipeline.conf.tuning.gt_tuning import (
    create_gt_tuning_widget,
    get_default_gt_path,
    get_default_target_dir
)

gt_path = get_default_gt_path()
target_dir = get_default_target_dir()

print(f'Project root: {project_root}')
print(f'File Ground Truth: {gt_path.relative_to(project_root)}')
print(f'Cartella Target Set: {target_dir.relative_to(project_root)}')
```

```python
# ==========================================
# AVVIO INTERFACCIA INTERATTIVA DI TUNING
# ==========================================
# 1. Seleziona l'immagine dal menu a tendina o usa i tasti Prec/Succ.
# 2. Se l'immagine contiene più monete, seleziona quale moneta vuoi modificare.
# 3. Sposta i cursori cx, cy e raggio (o usa i pulsanti +/- 1px e +/- 5px) osservando il riquadro di Zoom a destra.
# 4. Clicca su 'Salva Modifiche su gt.csv' per salvare in modo permanente le nuove coordinate.

widget = create_gt_tuning_widget(gt_csv_path=gt_path, target_dir=target_dir)
display(widget)
```
